```python
import nltk

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

text = """Tokenization splits text into words, stop-word removal removes unnecessary common words,
and lemmatization converts words into their base form. These steps help clean and simplify text for NLP processing."""

print(text)
```

```python
# 1. Sentence Tokenization

sentences = sent_tokenize(text)

print("Sentence Tokenization:")
for sentence in sentences:
    print(sentence)
```

```python
# 2. Word Tokenization

words = word_tokenize(text)

print("Word Tokenization:")
print(words)
```

```python
# Create dictionary
words = text.lower().split()
dictionary = {word: word for word in words}

print(dictionary)
```

```python
# 3. Stop-word Removal

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print("After Stop-word Removal:")
print(filtered_words)
```

```python
# 4. Lemmatization

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word.lower())
    for word in filtered_words
    if word.isalpha()
]

print("After Lemmatization:")
print(lemmatized_words)
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

trainer = BpeTrainer(vocab_size=100)
tokenizer.train_from_iterator([text], trainer)

print(tokenizer.encode(text).tokens)
```
