```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = """Natural Language Processing is an important field of Artificial Intelligence.
It helps computers understand and process human language."""

word_tokens1 = word_tokenize(text)

print("Word Tokens:")
print(word_tokens1)
```

```python
sentence_tokenize = sent_tokenize(text)
print(sentence_tokenize)
```

```python
dictionary = [
    ("Jammu","Kashmir"),
    ("Rajguru","Yogesh","Chiwate"),
    ("Satara","Maharashtrs")
] 
dictionary_tokenization = [word for word in word_tokens1 if word in dictionary]
```
