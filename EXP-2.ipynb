```python
import pandas as pd 
import pickle

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
df = pd.read_csv("./data/IMDB_Cleaned.csv")
df.head()
```

```python
#Generate the BOW
bow_vectorizers = CountVectorizer(max_features=5000)
x_bow = bow_vectorizers.fit_transform(df["clean_review"])
```

```python
x_bow
```

```python
x_bow.toarray()
```

```python
bow_vectorizers.get_feature_names_out()
```

```python
bow_df = pd.DataFrame(
    x_bow.toarray(),
    columns=bow_vectorizers.get_feature_names_out()
)
bow_df.head()
```

```python
#generate tfidf 
tfidf = TfidfVectorizer(max_features=5000)
tfidf
```

```python
x_tfidf = tfidf.fit_transform(df["clean_review"])
x_tfidf
```

```python
x_tfidf.toarray()
```

```python
tfidf.get_feature_names_out()
```

```python
tfidf_df = pd.DataFrame(
    x_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)
tfidf_df.head()
```

```python
#save features to csv files
bow_df.to_csv("./data/IMDB_BOW.csv",index=False)
tfidf_df.to_csv("./data/IMDB_TFIDF.csv",index=False)
print("Files saved sucessfully")
```

```python
#Save Vectorizer objects
with open("./assets/bow_vectorizer.pkl","wb") as file:
    pickle.dump(bow_vectorizers, file)

    
with open("./assets/tfidf_vectorizer.pkl","wb") as file:
    pickle.dump(tfidf, file)

print("Vectorizer objects saved sucessfully")
```
