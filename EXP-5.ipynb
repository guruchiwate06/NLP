```python
import re
import nltk 
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
#Rule based POS tagging Function
def rule_based_pos(word):
    word_lower = word.lower()
    #Pronun
    if word_lower in ["i","you","he","she","we","they","mr","him","her","us","them"]:
        return "Pronoun"
    #Determiners
    elif word_lower in ["the","a","an","this","that","these","those"]:
        return "Determiner"
    #Common verb 
    elif word_lower in ["is","am","are","was","were","be","been","being"]:
        return "Verb"
    #Modal verb
    elif word_lower in ["can","could","will","would","shall","should","may","might","must"]:
        return "Modal Verb"
    #Conjusction 
    elif word_lower in ["and","or","but","because","although","while"]:
        return "Conjuction"
    #Preposition
    elif word_lower in ["in","on","at","by","for","with","from","to","of"]:
        return "Preposition"
    #Adverb
    elif re.match(r".*ly$",word_lower):
        return "Adverb"

    #Adjective
    elif re.match(
        r".*(ous|ful|able|al|ive|less|ic|ish)$",
        word_lower
    ):
        return "Adjective"
    #Present participle 
    elif re.match(r".*ing$",word_lower):
        return "Present Participle"
    #Past tense
    elif re.match(r".*ed$",word_lower):
        return "Past tense Verb"
    #Number
    elif re.match(r".*^\d+$",word_lower):
            return "Number"
    #Plural noun
    elif re.match(r".*s$",word_lower):
            return "Plural Noun"
    #Punctuation
    elif re.match(r"^[,,!?;:]$",word_lower):
            return "Punctuation"
    #Default 
    else:
         return "Noun"

    
```
