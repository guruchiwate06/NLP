```python
from urllib.parse import urlparse, parse_qs
from youtube_transcript_api import YouTubeDataUnparsable
from gtts import gTTS

#Steo 1: Enter the YOutube Url copied from the browser
video_url = input("").strip()

video_id = video_url.split("v=")[1].split("&")[0]

print("Video ID",video_id)

#Step 3: Fetch the English teanscript
transcript = YoutubeTranscriptApi().fetch(
    video_id,
    language=["en"]
)
#Step 4: Combine transcript segments int text 
text = " ".join(segment.text for segment in transcript)
print("Transcript fetched sucessfully")
print("Transcript--------:\n",text)

#Step 5: Generate audio using Google Text-to-Speech
tts = gTTS(text=text, lang="en")

#Step 6: Save the audio as an MP3 file
output_file = "youtube_audio.mp3"
tts.save(output_file)


```
